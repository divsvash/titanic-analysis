# Titanic Dataset Analysis
Python & Data Analysis assignment

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# load the titanic dataset that comes with seaborn
df = sns.load_dataset("titanic")
```

## 1. First 10 records

```python
df.head(10)
```

## 2. Number of rows and columns

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 3. Data types of all variables

```python
df.dtypes
```

## 4. Missing values in each column

```python
df.isnull().sum()
```

## 5. Mean age of passengers

```python
print("Mean age:", round(df["age"].mean(), 2))
```

## 6. Number of male and female passengers

```python
df["sex"].value_counts()
```

## 7. Survival rate

```python
survival_rate = df["survived"].mean() * 100
print("Survival rate:", round(survival_rate, 2), "%")
```

## 8. Survival by gender

```python
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of passengers")
plt.legend(["did not survive", "survived"])
plt.show()
```

## 9. Histogram of passenger ages

```python
sns.histplot(df["age"], bins=30)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

## 10. Correlation between numerical variables

```python
df.corr(numeric_only=True).round(2)
```
